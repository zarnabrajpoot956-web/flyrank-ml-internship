## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Search question: Which visible pages are underperforming their position
tier's expected CTR, and deserve a metadata/title review?

Unit of analysis: a page (per client, per time window).

Output: a ranked list of pages, sorted by how far below their tier's
expected CTR they sit (the "gap"), with a reason code (e.g., "high
impressions, low CTR, strong position").

Action someone takes: a content editor reviews and rewrites the page's
title tag and meta description to better match search intent.

Cost of a wrong recommendation:
- False positive (flagged, but nothing was wrong): editor spends
  30-60 minutes rewriting a title that didn't need it — wasted time,
  low severity.
- False negative (missed a real underperformer): a page keeps
  under-capturing clicks — a slow, silent loss that compounds over time.
  
Why data/ML helps: a human can't manually compare thousands of pages
against tier-specific expected CTR at scale — this needs grouping and
averaging across the dataset, which code handles far better than
manual review.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os

print(os.getcwd())
print(os.listdir())
```

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import glob

matches = glob.glob("**/content_refresh_anonymized.csv", recursive=True)
print(matches)
df = pd.read_csv(matches[0])

visible = df[df["impressions_90d"] >= 100]

ctr_by_tier = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print("Mean CTR by position tier:")
print(ctr_by_tier.round(4))

top_tier = visible[visible["position_tier"].isin(["page_1", "top_3"])]
median_ctr = top_tier["ctr"].median()
underperformers = top_tier[top_tier["ctr"] < median_ctr]
print(f"\nPages in top tiers with below-median CTR: {len(underperformers)} of {len(top_tier)}")

print("\nContent type breakdown among underperformers:")
print(underperformers["content_type"].value_counts())
```

These numbers support Lane 4 as a worthwhile focus. CTR changes significantly by position tier, so pages should be compared only with other pages in the same tier. Around half of the top-ranked pages still have below-median CTR, showing there are many pages that could be improved. Most underperforming pages are keyword articles, although this may simply be because keyword articles are the most common content type. More analysis is needed before concluding that this content type performs worse than others.
